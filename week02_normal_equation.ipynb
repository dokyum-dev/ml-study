# Week 2 — 정규방정식으로 선형회귀 풀기

NumPy만으로 (XᵀX)⁻¹Xᵀy 를 구현.

**알게 된 것**: 절편 추정치가 참값에서 크게 벗어났는데,
데이터가 x=10~100 구간에만 있어서 x=0에서의 값은 외삽이기 때문.
β₀와 β₁의 상관계수가 -0.92로, 서로 보상하며 흔들린다.

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)
n = 50
ad = rng.uniform(10, 100, n)
sales = 3.5 * ad + 20 + rng.normal(0, 25, n)

plt.scatter(ad, sales)
plt.xlabel("ad spend")
plt.ylabel("sales")
plt.show()

print(ad.shape, sales.shape)
```

```python
X = np.column_stack([np.ones(n), ad])
y = sales

print(X.shape)
print(X[:3])
```

```python
beta = (np.linalg.inv(X.T@X) )@X.T@y
beta
```
